```python
name = input("Enter your name: ")
goal = input("Enter your carrier goal: ")
print("\nWelcome,", name)
print("Carrer Goal:", goal)
```

```python
python_skill = input("Do you know Python?(yes?no):")
numpy_skill = input("Do you know Numpy?(yes?no):")
pandas_skill = input("Do you know Pandas?(yes?no):")
sql_skill = input("Do you know sql?(yes?no):")
ml_skill = input("Do you know Machine learning?(yes?no):")
```

```python
import numpy as np
```

```python
skills = np.array([
    20 if python_skill.lower() == "yes"
else 0,
    20 if numpy_skill.lower() == "yes"
else 0,
    20 if pandas_skill.lower() == "yes"
else 0,
    20 if sql_skill.lower() == "yes"
else 0,
    20 if ml_skill.lower() == "yes"
else 0,  
])

print("Skill Scores:", skills)
print("Totall Score:",np.sum(skills))
```

```python
score = np.sum(skills)

if score <= 20:
    level = "Beginner"
if score <= 60:
    level = "Intermediate"
if score <= 80:
    level = "Adavanced"
else:
    level = "Expert"

print("\nYour Score:",score)
print("your Level:",level)
```

```python
print("\n==== AI Recommendation ====")

if pandas_skill.lower() == "no":
    print("📚 Learn Pandas")
if sql_skill.lower() == "no":
    print("📔Learn SQL")
if ml_skill.lower() == "no":
    print("🤖Learn Machine Learning") 
print("✅Keep practicing Python and Numpy")
```
